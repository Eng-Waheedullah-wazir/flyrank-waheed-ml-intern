## 2. Discover a real truth yourself

The safest, most satisfying early wins are **things you find in the data** — un-leakable, and they *are* the core lesson. Run the three cells below. Each is ~10 lines of pandas and each overturns a common SEO belief.

Every number is **computed live from the shipped CSV** — nothing is hardcoded.

```python
import pandas as pd, numpy as np
df = pd.read_csv("data/raw/content_refresh_anonymized.csv")
print(df.shape[0], "rows,", df.shape[1], "columns")
df.head(3)
```

### Discovery A — "High search volume means more traffic." Does it?

```python
corr = df["search_volume"].corr(df["impressions_90d"])
print(f"Correlation between search_volume and impressions_90d: {corr:.3f}")
print("Near zero -> keyword search volume barely predicts the traffic a page actually gets.")
```

### Discovery B — the CTR cliff by position
Click-through rate is not flat: it collapses as you move down the results.

```python
visible = df[df["impressions_90d"] >= 100]
ctr_by_pos = visible.groupby("position_tier")["ctr"].mean().sort_values(ascending=False)
print(ctr_by_pos.round(4).to_string())
ctr_by_pos.plot(kind="bar", title="Mean CTR by position tier (impressions >= 100)", ylabel="CTR");
```

### Discovery C — is longer content the lever?
Compare word count for **declining** vs **growing** pages.

```python
wc = df.groupby("trend_direction")["word_count"].median()
print(wc.round(0).to_string())
print("\n'down' vs 'up' pages have almost the same median word count -> length is not the lever.")
```

```python
# Your discovery here
# Keep only pages that actually received impressions
visible = df[df["impressions_90d"] > 0]

corr = visible["search_volume"].corr(visible["impressions_90d"])

print(f"Correlation (impressions > 0): {corr:.3f}")
```

After removing pages with zero impressions, the correlation remained very weak. This suggests that keyword search volume alone is not a good predictor of how many impressions a page actually receives





```python
table = (
    df.pivot_table(
        values="ctr",
        index="position_tier",
        columns="content_type",
        aggfunc="mean"
    )
)

print(table.round(4))
```

**Research Question:** Which content type has the lowest CTR within the same Google ranking position?

**Method:** I grouped the data by position_tier and content_type and calculated the mean CTR for each combination.

**Finding:**Comparison articles consistently have the lowest CTR across almost every position tier. For example, on Page 1 their average CTR is 0.1319, compared to 0.4485 for keyword articles. In the Top 3 positions, comparison articles even have a mean CTR of 0.0000, while keyword articles average 0.9362.

**Conclusion:** Ranking position alone does not determine CTR. The type of content also influences user clicks. Comparison articles appear to attract fewer clicks than keyword articles at the same ranking positions.

```python
corr = df["avg_position"].corr(df["ctr"])

print(f"Correlation between avg_position and CTR: {corr:.3f}")
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(7,5))
plt.scatter(df["avg_position"], df["ctr"], alpha=0.2)
plt.xlabel("Average Position")
plt.ylabel("CTR")
plt.title("Average Position vs CTR")
plt.show()
```

CTR decreases as average ranking position becomes worse. Pages ranking closer to Position 1 receive substantially higher click-through rates

```python
age = (
    df.groupby("trend_direction")["content_age_days"]
      .median()
      .sort_values()
)

print(age)
```

**Discovery:** I investigated whether content age affects page trends by comparing the median content_age_days across trend_direction. The results showed that down pages had the lowest median age (216 days), while stable pages had the highest (300 days). This suggests that content age alone is not a reliable indicator of whether a page will grow or decline
